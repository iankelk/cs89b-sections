### POS Tagging and Person Named Entity Recognition (NER) using NLTK

```python
import nltk
import re
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk import pos_tag

# Ensure NLTK packages are downloaded
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt_tab')

# Sample text (news article) from New York Times (https://www.nytimes.com/2024/11/22/business/elon-musk-tesla-china.html)
news_article = """
From electric cars to solar panels, Mr. Musk has built businesses in high-tech manufacturing sectors now targeted by Beijing for Chinese dominance.
No American business leader has more visibly and lavishly supported President-elect Donald J. Trump than Elon Musk — and few if any have a more complex relationship with China, a country that Mr. Trump has vowed to confront with higher tariffs and other measures.
Mr. Musk has a lot on the line. His best-known company, the electric vehicle maker Tesla, makes half its cars in China. Tesla sells more cars in China than anywhere except the United States, and his local competition is getting stronger. Chinese regulators have not yet allowed Tesla to offer its latest assisted-driving and self-driving car technology, while allowing Chinese automakers to race ahead with similar systems.
Mr. Musk has personally appealed to China’s premier, Li Qiang, for permission to proceed with what Tesla calls Full Self-Driving as the company’s market share in China has dwindled. Some experts have suggested that Beijing may be able to turn Mr. Musk into an influential ally in trying to persuade Mr. Trump to take a more conciliatory approach on trade.
“When Chinese leaders have an important message to convey to President Trump, Elon Musk would clearly be the best conduit,” said Michael Dunne, a longtime China automotive consultant now based in San Diego.
"""

# Tokenize the text into sentences. Display the sentences.
sentences = sent_tokenize(news_article)
print("Original Sentences:")
for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence}")

# Tokenize each sentence into words. Display the tokenized sentences
tokenized_sentences = [word_tokenize(sentence) for sentence in sentences]
print("\nTokenized Sentences and Words:")
for i, tokenized_sentence in enumerate(tokenized_sentences, start=1):
    print(f"Sentence {i}: {tokenized_sentence}")

# Perform POS tagging on each tokenized sentence. Display the POS tagged sentences.
tagged_sentences = [pos_tag(sentence) for sentence in tokenized_sentences]
print("\nPOS Tagged Sentences:")
for i, tagged_sentence in enumerate(tagged_sentences, start=1):
    print(f"Sentence {i}: {tagged_sentence}")

# Function to extract unique entities based on POS patterns and regular expressions
def extract_entities(tagged_sentences):
    entities = {'PERSON': set()}  # Use a set to ensure uniqueness

    for sentence in tagged_sentences:
        sentence_text = ' '.join([word for word, tag in sentence])

        # Rule for detecting Person Names with titles
        matches = re.compile(r'\b(?:Mr|Mrs|Ms|Dr|Prof)\.?\s+(?:[A-Z][a-z]+\s+)?[A-Z][a-z]+\b').findall(sentence_text)
        entities['PERSON'].update(matches)  # Update the set with new matches

    return entities

# Apply rules to extract entities
extracted_entities = extract_entities(tagged_sentences)

# Display recognized entities
print("\nRecognized Entities (Unique):")
for entity_type, entity_set in extracted_entities.items():
    print(f"{entity_type}: {list(entity_set)}")  # Convert the set to a list for display
```

### Named Entity Recognition (NER) with spaCy

```python
import spacy
from spacy import displacy

# Load the default SpaCy model
nlp = spacy.load("en_core_web_sm")

# Sample text (news article) from New York Times (https://www.nytimes.com/2024/11/22/business/elon-musk-tesla-china.html)
news_article = """
From electric cars to solar panels, Mr. Musk has built businesses in high-tech manufacturing sectors now targeted by Beijing for Chinese dominance.
No American business leader has more visibly and lavishly supported President-elect Donald J. Trump than Elon Musk — and few if any have a more complex relationship with China, a country that Mr. Trump has vowed to confront with higher tariffs and other measures.
Mr. Musk has a lot on the line. His best-known company, the electric vehicle maker Tesla, makes half its cars in China. Tesla sells more cars in China than anywhere except the United States, and his local competition is getting stronger. Chinese regulators have not yet allowed Tesla to offer its latest assisted-driving and self-driving car technology, while allowing Chinese automakers to race ahead with similar systems.
Mr. Musk has personally appealed to China’s premier, Li Qiang, for permission to proceed with what Tesla calls Full Self-Driving as the company’s market share in China has dwindled. Some experts have suggested that Beijing may be able to turn Mr. Musk into an influential ally in trying to persuade Mr. Trump to take a more conciliatory approach on trade.
“When Chinese leaders have an important message to convey to President Trump, Elon Musk would clearly be the best conduit,” said Michael Dunne, a longtime China automotive consultant now based in San Diego.
"""

# Process the text with the default model
doc = nlp(news_article)

# Display recognized entities by the default model
print("Recognized Entities:")
for entity in doc.ents:
    print(f"Entity: {entity.text}, Type: {entity.label_}")

# Visualize the entities in the text using displacy
displacy.render(doc, style="ent", jupyter=True)
```

```python
from spacy.training import Example
import random

# Load a base model for auto-annotation
base_nlp = spacy.load("en_core_web_sm")

# Load the model we will actually fine-tune
nlp = spacy.load("en_core_web_sm")

# VEHICLE-only training data
TRAIN_DATA = [
    ("cars in China.", {"entities": [(0, 4, "VEHICLE")]}),
    ("Tesla has a lot on the line as an electric vehicle maker.", {"entities": [(43, 50, "VEHICLE")]}),
    ("The electric vehicle Tesla is made in China.", {"entities": [(4, 20, "VEHICLE")]}),
    ("The car is parked outside.", {"entities": [(4, 7, "VEHICLE")]}),
    ("I just bought a new electric car.", {"entities": [(20, 32, "VEHICLE")]}),
    ("He drives a hybrid vehicle.", {"entities": [(12, 26, "VEHICLE")]}),
    ("Toyota is one of the biggest car manufacturers.", {"entities": [(29, 32, "VEHICLE")]}),
    ("My family loves our Honda Civic, it's a great car.", {"entities": [(20, 31, "VEHICLE"), (46, 49, "VEHICLE")]}),
    ("Ford trucks dominate the market.", {"entities": [(5, 11, "VEHICLE")]}),
    ("They unveiled the latest model of their sports car.", {"entities": [(40, 50, "VEHICLE")]}),
]

# Enrich TRAIN_DATA with spaCy's existing entities + VEHICLE
vehicle_train_data = TRAIN_DATA
TRAIN_DATA = []

for text, ann in vehicle_train_data:
    # Get spaCy's current entities on this sentence
    doc = base_nlp(text)
    base_ents = [(ent.start_char, ent.end_char, ent.label_) for ent in doc.ents]

    # Our manually defined VEHICLE entities
    vehicle_ents = ann["entities"]

    # Helper to detect span overlap
    def overlaps(span1, span2):
        s1, e1 = span1
        s2, e2 = span2
        return not (e1 <= s2 or e2 <= s1)

    # Keep base entities that do not overlap any VEHICLE span
    merged_ents = []
    for start, end, label in base_ents:
        if not any(overlaps((start, end), (v_start, v_end))
                   for (v_start, v_end, v_label) in vehicle_ents):
            merged_ents.append((start, end, label))

    # Add VEHICLE entities on top
    merged_ents.extend(vehicle_ents)

    TRAIN_DATA.append((text, {"entities": merged_ents}))

# Fine-tune NER with the enriched TRAIN_DATA
ner = nlp.get_pipe("ner")
if "VEHICLE" not in ner.labels:
    ner.add_label("VEHICLE")

examples = []
for text, ann in TRAIN_DATA:
    doc = nlp.make_doc(text)
    examples.append(Example.from_dict(doc, ann))

other_pipes = [p for p in nlp.pipe_names if p != "ner"]
with nlp.disable_pipes(*other_pipes):
    optimizer = nlp.resume_training()
    for _ in range(60):  # you can reduce to 20 if you want a lighter update
        random.shuffle(examples)
        losses = {}
        nlp.update(examples, sgd=optimizer, drop=0.2, losses=losses)
        # print(losses)  # optional

# Check if the model learned VEHICLE on the toy sentences
print("Check on training sentences:")
for text, _ in vehicle_train_data:
    doc_train = nlp(text)
    print(text)
    print([(ent.text, ent.label_) for ent in doc_train.ents])
    print()

# Run on the news article
doc = nlp(news_article)
print("Recognized Entities with Custom Entity:")
for entity in doc.ents:
    print(f"Entity: {entity.text}, Type: {entity.label_}")

displacy.render(doc, style="ent", jupyter=True)
```
